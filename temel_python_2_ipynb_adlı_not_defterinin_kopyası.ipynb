## 5️⃣ Döngüler (for / while)

```python
# Örnek: enumerate ve zip
renkler = ["kırmızı", "yeşil", "mavi"]

for i, renk in enumerate(renkler, start=1):
    print(f"{i}. {renk}")

print("---")

isimler = ["Ali", "Ayşe", "Can"]
notlar = [85, 92, 78]

for isim, not_ in zip(isimler, notlar):
    print(f"{isim}: {not_}")
```

### 🔸 Soru 10:
`while` döngüsü ile basamak sayısı bulma: Verilen bir sayının kaç basamaklı olduğunu, sayıyı 10'a bölerek bulun (string dönüşümü kullanmadan).

```python
# Cevabınızı buraya yazın:
# Örnek: 4752 -> 4 basamaklı
sayi = 4752
adet = 0

while sayi > 0:
    sayi //= 10
    adet += 1

print(adet)
```

### 🔸 Soru 12:
Bir öğrenci listesi var. Her öğrenci `(isim, not)` tuple'ı olarak saklanıyor.
- Notu 70'in üzerinde olanları filtreleyin
- En yüksek notu alan öğrenciyi bulun
- Notların ortalamasını hesaplayın

```python
# Cevabınızı buraya yazın:
ogrenciler = [
    ("Ahmet", 85),
    ("Büşra", 62),
    ("Can", 91),
    ("Deniz", 74),
    ("Elif", 58),
    ("Furkan", 88)
]

gecenler = [ogrenci for ogrenci in ogrenciler if ogrenci[1] > 70]
print("70 üstü alanlar:", gecenler)

en_yuksek = max(ogrenciler, key=lambda x: x[1])
print("En yüksek not alan:", en_yuksek)

toplam = sum(notu for isim, notu in ogrenciler)
ortalama = toplam / len(ogrenciler)
print("Not ortalaması:", ortalama)
```

### 🔸 Soru 13 (Set):
İki sınıfın öğrenci listesi var. `set` kullanarak:
- Her iki sınıfta da bulunan öğrencileri bulun (kesişim)
- Toplam kaç farklı öğrenci olduğunu bulun (birleşim)
- Sadece A sınıfında olup B'de olmayan öğrencileri bulun (fark)

```python
# Cevabınızı buraya yazın:
sinif_a = {"Ali", "Ayşe", "Can", "Dilan", "Ece"}
sinif_b = {"Can", "Dilan", "Fatih", "Gökçe", "Hakan"}

print("kesişim:", sinif_a & sinif_b)
print("birleşim:", len(sinif_a | sinif_b))
print("fark:", sinif_a - sinif_b)
```

### 🔸 Soru 14:
Bir cümledeki her harfin kaç kez geçtiğini sayan bir sözlük oluşturun (boşlukları saymayin, büyük/küçük harf farkı olmasın).

```python
# Cevabınızı buraya yazın:
cumle = "merhaba dünya"

harf_sayilari = {}

cumle = cumle.lower()   # Büyük/küçük harf farkını kaldır

for harf in cumle:
    if harf != " ":    # Boşlukları sayma
        harf_sayilari[harf] = harf_sayilari.get(harf, 0) + 1

print(harf_sayilari)
```

## 8️⃣ Fonksiyonlar

```python
# Örnek: *args ve **kwargs
def toplam(*sayilar):
    return sum(sayilar)

print(toplam(1, 2, 3))
print(toplam(10, 20, 30, 40, 50))

def kullanici_bilgi(**bilgiler):
    for anahtar, deger in bilgiler.items():
        print(f"{anahtar}: {deger}")

kullanici_bilgi(ad="Selin", yas=25, sehir="İzmir")
```

### 🔸 Soru 15:
Özyinelemeli (recursive) bir fonksiyon yazın: `faktoriyel(n)` - n! hesaplasın.
- 0! = 1
- n! = n × (n-1)!

5!, 10! değerlerini hesaplayın.

```python
# Cevabınızı buraya yazın:
def faktoriyel(n):
    if n == 0:
        return 1
    return n * faktoriyel(n - 1)

print(faktoriyel(5))
print(faktoriyel(10))
```

### 🔸 Soru 16:
`lambda` fonksiyonu kullanarak:
- Bir sayının karesini hesaplayan lambda
- İki sayının büyüğünü döndüren lambda
- `sorted()` ile bir isim listesini son harfe göre sıralayan lambda

```python
# Cevabınızı buraya yazın:
isimler = ["Zeynep", "Ali", "Mahmut", "Su", "Berk"]

kare = lambda x: x ** 2
buyuk = lambda a, b: a if a > b else b

print(kare(9))
print(buyuk(5, 11))

sirali = sorted(isimler, key=lambda x: x[-1])

print(sirali)
```

### 🔸 Soru 17:
Bir alışveriş listesini `alisveris.txt` dosyasına yazın (her ürün ayrı satırda). Sonra dosyayı okuyarak ürünleri numaralı şekilde ekrana yazdırın.

### 🔸 Soru 18 (Bütünleştirici):
Basit bir **öğrenci not sistemi** yazın:

- `ogrenciler` adında boş bir sözlük oluşturun (anahtar: isim, değer: notlar listesi)
- `ogrenci_ekle(isim)`: Yeni öğrenci ekler
- `not_ekle(isim, not_)`: Öğrenciye not ekler. Öğrenci yoksa hata mesajı verir.
- `ortalama_hesapla(isim)`: Öğrencinin not ortalamasını döndürür
- `sinif_raporu()`: Tüm öğrencileri ve ortalamalarını yazdırır, sınıf ortalamasını gösterir

En az 3 öğrenci ve her birine 3 not girerek test edin.

```python
# Cevabınızı buraya yazın:
ogrenciler = {}

def ogrenci_ekle(isim):
     ogrenciler[isim] = []

def not_ekle(isim, not_):
    if isim in ogrenciler:
        ogrenciler[isim].append(not_)
    else:
        print("Öğrenci yok")

def ortalama_hesapla(isim):
    return sum(ogrenciler[isim]) / len(ogrenciler[isim])

def sinif_raporu():
    toplam = 0

    for isim in ogrenciler:
        ort = ortalama_hesapla(isim)
        toplam += ort
        print(isim, ":", ort)

    print("Sınıf Ortalaması:", toplam / len(ogrenciler))


# Test kodu:
ogrenci_ekle("Ahmet")
ogrenci_ekle("Büşra")
ogrenci_ekle("Can")
not_ekle("Ahmet", 85)
not_ekle("Ahmet", 90)
not_ekle("Ahmet", 78)
not_ekle("Büşra", 92)
not_ekle("Büşra", 88)
not_ekle("Büşra", 95)
not_ekle("Can", 70)
not_ekle("Can", 65)
not_ekle("Can", 73)
sinif_raporu()
```
